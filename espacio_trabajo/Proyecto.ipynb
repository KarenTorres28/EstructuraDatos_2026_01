```python
def Notacion_Polaca(A):
    S = ArrayStack()     

  
    if isinstance(A, str):
        tokens = A.split()
        for t in tokens:
            if t in "+-*/^":
                if len(S) < 2:
                    raise ValueError("No hay suficientes operandos")
                b = S.pop()
                a = S.pop()
                if t == "+": 
                  r = a + b
                elif t == "-": 
                  r = a - b
                elif t == "*": 
                  r = a * b
                elif t == "/": 
                  r = a / b
                elif t == "^": 
                  r = a ** b
                S.push(r)
            else:
                try:
                    S.push(float(t))
                except:
                    raise ValueError(f"Token inválido: {t}")

   
    elif isinstance(A, ArrayStack):
        n = len(A)
        aux = ArrayStack()

       
        for i in range(n):
            aux.push(A.pop())

      
        for i in range(n):
            token = aux.pop()
            A.push(token)  

            if token in "+-*/^":
                if len(S) < 2:
                    raise ValueError("No hay suficientes operandos")
                b = S.pop()
                a = S.pop()
                if token == "+": 
                  r = a + b
                elif token == "-": 
                  r = a - b
                elif token == "*": 
                  r = a * b
                elif token == "/": 
                  r = a / b
                elif token == "^": 
                  r = a ** b
                S.push(r)
            else:
                try:
                    S.push(float(token))
                except:
                    raise ValueError(f"Token inválido: {token}")

    else:
        raise TypeError("Tipo de entrada inválido")

    if len(S) != 1:
        raise ValueError("La expresión termina con más de un número en la pila")

    return S.pop()

##Punto Extra:
def Infijo_a_Postfijo(exp):
    S = ArrayStack()
    salida = []

    exp = exp.replace("(", " ( ").replace(")", " ) ")
    for op in "+-*/^":
        exp = exp.replace(op, f" {op} ")

    tokens = exp.split()
    prec = {"+":1, "-":1, "*":2, "/":2, "^":3}

    for t in tokens:
        if t == "(":
            S.push(t)

        elif t == ")":
            n = len(S)
            for a in range(n):
                if S.top() == "(":
                    S.pop()
                    break
                salida.append(S.pop())

        elif t in "+-*/^":
            n = len(S)
            for e in range(n):
                if S.is_empty() or S.top() == "(" or prec[S.top()] < prec[t]:
                    break
                salida.append(S.pop())
            S.push(t)

        else:
            salida.append(t)

    n = len(S)
    for h in range(n):
        salida.append(S.pop())

    return " ".join(salida)
```
